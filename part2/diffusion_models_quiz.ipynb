# Diffusion Models Quiz

Multiple-choice questions based on the article [What are Diffusion Models?](https://lilianweng.github.io/posts/2021-07-11-diffusion-models/).

**How to use:** pick one option per question, then press **Check Answers**.


```python
from __future__ import annotations

from dataclasses import dataclass
from typing import List, Dict, Any


@dataclass
class Q:
    prompt: str
    choices: List[str]
    answer_idx: int
    explanation: str


QUIZ: List[Q] = [
    Q(
        prompt="In the DDPM forward process, what is q(x_t | x_{t-1})?",
        choices=[
            "A Gaussian with mean sqrt(1 - beta_t) * x_{t-1} and variance beta_t * I",
            "A Bernoulli distribution over {0,1}",
            "A deterministic mapping x_t = x_{t-1} + beta_t * epsilon",
            "A Gaussian with mean x_{t-1} and fixed variance I (independent of t)",
        ],
        answer_idx=0,
        explanation="DDPM defines q(x_t | x_{t-1}) = N(x_t; sqrt(1 - beta_t) x_{t-1}, beta_t I).",
    ),
    Q(
        prompt="Define alpha_t and bar{alpha}_t in terms of beta_t.",
        choices=[
            "alpha_t = 1 - beta_t, bar{alpha}_t = prod_{i=1..t} alpha_i",
            "alpha_t = beta_t, bar{alpha}_t = prod_{i=1..t} beta_i",
            "alpha_t = 1/beta_t, bar{alpha}_t = 1 - prod_{i=1..t} alpha_i",
            "alpha_t = beta_t - 1, bar{alpha}_t = sum_{i=1..t} alpha_i",
        ],
        answer_idx=0,
        explanation="The article uses alpha_t := 1 - beta_t and bar{alpha}_t := product_{i=1}^t alpha_i.",
    ),
    Q(
        prompt="Which statement best describes x_t as a function of x_0 and epsilon in DDPM?",
        choices=[
            "x_t = sqrt(bar{alpha}_t) x_0 + sqrt(1 - bar{alpha}_t) * epsilon, with epsilon ~ N(0, I)",
            "x_t = x_0 + bar{alpha}_t * epsilon, with epsilon ~ Uniform(0, 1)",
            "x_t = sqrt(beta_t) x_0 + sqrt(1 - beta_t) * epsilon, with epsilon ~ N(0, 1) but bar{alpha}_t is not needed",
            "x_t is always identical to x_0 for all t",
        ],
        answer_idx=0,
        explanation="A key property: x_t has a closed form using bar{alpha}_t and a standard Gaussian noise epsilon.",
    ),
    Q(
        prompt="The reverse conditional q(x_{t-1} | x_t, x_0) becomes tractable because it is…",
        choices=[
            "Gaussian with mean tilde{mu}_t(x_t, x_0) and variance tilde{beta}_t I",
            "Uniform on a hypersphere",
            "Student-t with heavy tails only (no closed form Gaussian term)",
            "Discrete categorical distribution over a finite codebook",
        ],
        answer_idx=0,
        explanation="Conditioned on x_0, the reverse conditional is Gaussian: q(x_{t-1} | x_t, x_0) = N(tilde{mu}_t, tilde{beta}_t I).",
    ),
    Q(
        prompt="In DDPM training, what is the simplified objective that the article describes?",
        choices=[
            "Predict epsilon: minimize E[||epsilon - epsilon_theta(x_t, t)||^2] (plus a constant)",
            "Maximize the log determinant of the Jacobian of the network",
            "Classify whether t is even or odd",
            "Minimize E[||x_t - x_0||^2] directly (without sampling epsilon)",
        ],
        answer_idx=0,
        explanation="The simplified DDPM loss turns into noise-prediction: MSE between the true epsilon and epsilon_theta.",
    ),
    Q(
        prompt="DDIM becomes deterministic when…",
        choices=["eta = 0", "beta_t = 0", "eta = 1", "t = 0"],
        answer_idx=0,
        explanation="DDIM uses a parameter eta controlling stochasticity; eta=0 yields deterministic sampling.",
    ),
    Q(
        prompt="Classifier-free guidance (CFG) works by…",
        choices=[
            "Training with random condition dropping so the model learns both conditional and unconditional scores",
            "Only training a conditional classifier f_phi and using gradients during sampling",
            "Replacing the diffusion process with an autoregressive model",
            "Training separate unconditional and conditional networks that never share parameters",
        ],
        answer_idx=0,
        explanation="CFG trains one network with periodic condition dropping, so epsilon_theta(x_t, t, y) and epsilon_theta(x_t, t) are available from the same model.",
    ),
    Q(
        prompt="Which statement about variance schedules is consistent with the article?",
        choices=[
            "Cosine schedules are designed to provide a near-linear drop in the middle and subtle changes near the endpoints",
            "All schedules yield identical training curves by definition",
            "Schedules affect only sampling, not training",
            "Only linear schedules are valid (others are invalid beta_t)",
        ],
        answer_idx=0,
        explanation="The article contrasts linear/polynomial/cosine schedules and motivates cosine schedules for improved performance.",
    ),
    Q(
        prompt="A key idea of latent diffusion models (LDM) is that diffusion happens in…",
        choices=[
            "a learned latent space (z) rather than pixel space",
            "the frequency domain only (FFT magnitude) and never in spatial domain",
            "a discrete token space like a VQ-encoder only (no continuous latents)",
            "the same original image space, so LDM is only a minor refactor",
        ],
        answer_idx=0,
        explanation="LDM compresses images with an autoencoder into latents and performs diffusion/denoising in that latent space for efficiency.",
    ),
    Q(
        prompt="Which sampling speed-up is described as using fewer steps S by skipping diffusion updates?",
        choices=[
            "Fewer sampling steps / strided schedules (S < T)",
            "Increasing T to make sampling easier",
            "Removing the denoiser network entirely",
            "Sampling one timestep only (t=1 always)",
        ],
        answer_idx=0,
        explanation="The article discusses strided sampling: choose fewer timesteps tau_1 < ... < tau_S where S < T.",
    ),
]


def check_answers(selected: List[int]) -> Dict[str, Any]:
    """selected holds one integer index per question."""
    if len(selected) != len(QUIZ):
        raise ValueError(f"Expected {len(QUIZ)} answers, got {len(selected)}")

    correct = 0
    per_q: List[Dict[str, Any]] = []
    for i, (q, a) in enumerate(zip(QUIZ, selected)):
        ok = (a == q.answer_idx)
        correct += int(ok)
        per_q.append(
            {
                "i": i,
                "prompt": q.prompt,
                "picked": q.choices[a] if 0 <= a < len(q.choices) else str(a),
                "correct": q.choices[q.answer_idx],
                "ok": ok,
                "explanation": q.explanation,
            }
        )

    return {
        "correct": correct,
        "total": len(QUIZ),
        "score": correct / len(QUIZ),
        "per_q": per_q,
    }


TOTAL = len(QUIZ)
print(f"Quiz loaded: {TOTAL} questions")
```

```python
try:
    import ipywidgets as widgets
    from IPython.display import display, clear_output

    ui_radios = []
    for idx, q in enumerate(QUIZ):
        # Each option is (value, label) so rb.value is the tuple.
        rb = widgets.RadioButtons(
            options=list(enumerate(q.choices)),
            description=f"Q{idx+1}",
            value=None,
            disabled=False,
            layout=widgets.Layout(width="100%"),
        )
        ui_radios.append(rb)

    check_btn = widgets.Button(
        description="Check Answers",
        button_style="success",
        tooltip="Evaluate your selections",
    )

    out = widgets.Output()

    def on_check_clicked(_):
        with out:
            clear_output()
            try:
                selected = []
                for rb in ui_radios:
                    if rb.value is None:
                        raise ValueError("Please answer all questions before checking.")
                    val = rb.value
                    if isinstance(val, (tuple, list)):
                        selected.append(int(val[0]))
                    else:
                        selected.append(int(val))

                res = check_answers(selected)
                print(f"Score: {res['correct']} / {res['total']} ({res['score']*100:.1f}%)")
                print("\nPer-question feedback:")
                for item in res["per_q"]:
                    status = "✅" if item["ok"] else "❌"
                    print(f"\n{status} Q{item['i']+1}: {item['prompt']}")
                    print(f"  Your answer: {item['picked']}")
                    print(f"  Correct:     {item['correct']}")
                    print(f"  Explanation: {item['explanation']}")
            except Exception as e:
                print(str(e))

    check_btn.on_click(on_check_clicked)
    display(*ui_radios, check_btn, out)

except Exception as e:
    print("ipywidgets is not available. You can still check answers by calling:")
    print("  check_answers(selected)")
    print(f"Details: {e}")
    selected = [0 for _ in range(len(QUIZ))]
    print("Example selected indices initialized to zeros.")
```
